# Two measured windows and a four-product decode

**Public circuit improvements compose when their actual interfaces agree.**
This version integrates Agniv Banerjee / SomeJoint's two HMR-crossing inverse
windows, Syed Asad Ali's five-product-to-four-product coefficient decode and
306/292-AND angle loaders, Isabe11aM00re's fixed-word weight refit, and
Yoyo's one-CCZ decoder cache edit. Their circuit and refit gains belong to
those authors. Our new work is the hash-pinned composition, independent Mac
MPS audits with negative controls, and two new complete original native checks.

Four Loaders V5 has official Public Score **24.50427**. At this study's
checkpoint Syed Own V2 and orecord V10 shared the observed public maximum
**24.53786**. The new assembly below has no official result at authoring time;
its local geometric mean is labelled separately. Exact source versions,
bytes, scientific conditions, full receipts and limitations are retained.


## 1. What is being composed

Inverse: retain SomeJoint's normalization and phase-split base and its two
measured-product reuse windows, then replace one exact coefficient-decode
span with Syed V2's public 5-to-4 resynthesis. The slice contains only
X/CX/CCX: one fewer paid CCX, 138 more CX and six more X. The old span must
occur exactly once; every source, slice, output and SDK roundtrip is hashed.

Chemistry: Syed V2's compiler outputs use 306 CCX at batches 0–3 and 292 at
batches 4–14. One forward pass costs **4×306 + 11×292 = 4,436**; the two
passes cost 8,872. The complete measured chemistry bill is 15,525 Toffoli at
428 qubits. Isabe's weights use the same U/word/tree/brot data, so their
exact factor bytes can accompany these loaders. The compiler and weights
were published by Syed and Isabe; we use their compiled/data outputs.

DQI: retain the exact Yoyo `06aa…` binary that already passed our complete
102,400-shot original native check. The decoder is approximate and its
repetition penalty remains in the metric. We do not repeat that native check.

## 3. Reachable loaders and measurement phases

A loader changes the preceding five words into the next five words, and
performs the reverse on the return path, for all 1,363 valid addresses.
These sign-copy words satisfy bit 7 = bit 6. This is the reachable contract
clarified by [host Tanuj Khattar](https://www.kaggle.com/competitions/google-verifiable-quantum-circuit-compilation/discussion/753092),
rather than a general arbitrary-target XOR claim.

For HMR result m, erasing a computed value f adds phase (-1)^(m f).
We represent each measurement result by its own formal variable and XOR
every conditional CZ contribution into that variable's coefficient. The
final constant phase and every coefficient must vanish for each reachable
input. Address and scratch restoration are checked separately; R is allowed
only on zero in this audit. This gives all-outcome phase coverage for this
narrow measurement-independent classical loader family.

Syed V2 adds balanced nonnested PUSH/CZ-only/POP blocks. Expanding each to
CZ conditioned on the same bit is exact because those blocks contain no
value update or measurement. Our first, narrower reader rejected opcode129;
the separate guarded-CZ reader retained that failure and explicitly widened
the accepted grammar. Native SDK exports below match the original streams,
before this normalization, binding the Mac proof to the actual binaries.


```python
def normalize_loader_cz(text):
    guard, normalized = None, []
    for line in text.splitlines():
        fields = line.split()
        if fields[0] in ("REGISTER", "APPEND_TO_REGISTER"):
            assert guard is None
            normalized.append(line)
        elif fields[0] == "PUSH_CONDITION":
            assert guard is None and len(fields) == 3 and fields[1] == "if"
            assert re.fullmatch(r"b\d+", fields[2])
            guard = fields[2]
        elif fields[0] == "POP_CONDITION":
            assert guard is not None and len(fields) == 1
            guard = None
        elif guard is not None:
            assert fields[0] == "CZ" and len(fields) == 3
            normalized.append(line + " if " + guard)
        else:
            assert fields[0] in ("X", "CX", "CCX", "R", "HMR", "CZ")
            if "if" in fields:
                assert fields[0] == "CZ" and len(fields) == 5 and fields[3] == "if"
            normalized.append(line)
    assert guard is None
    return "\n".join(normalized) + "\n"


loader_texts, original_loader_texts, export_rows = {}, {}, []
export_dir = WORK / "mps_loader_inputs"
export_dir.mkdir(exist_ok=True)
shutil.copyfile(submission / "dfthcsos/submission.npz", export_dir / "submission.npz")
for k in range(15):
    name = f"angle_k_{k}"
    output = export_dir / (name + ".kmx")
    km.Circuit.from_file(str(submission / "dfthcsos" / (name + ".kmb")), format="kmb").to_file(
        str(output), format="kmx"
    )
    assert sha256(output) == EXPECTED_LOADER_CANONICAL[str(k)], k
    original_loader_texts[name] = output.read_text()
    loader_texts[name] = normalize_loader_cz(output.read_text())
    (export_dir / (name + ".normalized.kmx")).write_text(loader_texts[name])
    export_rows.append(
        dict(
            batch=k,
            CCX=sum(l.startswith("CCX ") for l in loader_texts[name].splitlines()),
            canonical_sha256=sha256(output),
        )
    )
assert [row["CCX"] for row in export_rows] == [306] * 4 + [292] * 11
(export_dir / "angle_kmxs.json").write_text(json.dumps(loader_texts))
display(pd.DataFrame(export_rows))
```

```python
def run_symbolic_mps_audit(data_dir):
    """MPS proof for normalized narrow reachable loaders; not arbitrary-target XOR."""
    import os

    os.environ["OMP_NUM_THREADS"] = "1"
    os.environ["PYTORCH_ENABLE_MPS_FALLBACK"] = "0"
    import torch, numpy as np, json, re, time
    from pathlib import Path
    from collections import Counter

    torch.set_num_threads(1)
    assert torch.backends.mps.is_available()
    t = time.monotonic()
    dev = "mps"
    root = Path(data_dir)
    MASK = (1 << 63) - 1
    with np.load(root / "submission.npz") as z:
        words = torch.from_numpy(z["angle_words"].copy()).to(dev, dtype=torch.int64) & 255
    A, columns = words.shape
    N = A * 2
    W = (N + 62) // 63
    ids = torch.arange(W * 63, device=dev, dtype=torch.int64).reshape(W, 63)
    addr = ids % A
    cases = ids // A
    weights = torch.ones(63, device=dev, dtype=torch.int64) << torch.arange(
        63, device=dev, dtype=torch.int64
    )

    def pack(bits):
        return (bits * weights).sum(-1)

    sel = torch.stack([pack(addr >> j & 1) for j in range(11)])
    phase_coefficients = []
    phase_bits = {}
    FULL = torch.full((W,), MASK, device=dev, dtype=torch.int64)
    payload = json.loads((root / "angle_kmxs.json").read_text())
    records = []
    for k in range(15):
        text = payload["angle_k_" + str(k)]
        lines = text.splitlines()
        current = words[:, k * 5 : k * 5 + 5]
        previous = words[:, (k - 1) * 5 : k * 5] if k else torch.zeros_like(current)
        current = current[addr.reshape(-1)].reshape(W, 63, 5).permute(2, 0, 1)
        previous = previous[addr.reshape(-1)].reshape(W, 63, 5).permute(2, 0, 1)
        target = torch.where((cases % 2 == 0)[None], previous, current)
        target_bits = torch.stack([pack(target[j] >> b & 1) for j in range(5) for b in range(8)])
        q = torch.zeros((128, W), device=dev, dtype=torch.int64)
        q[:11] = sel
        q[11:51] = target_bits
        bits = {}
        phase = torch.zeros(W, device=dev, dtype=torch.int64)
        phase_coefficients = []
        phase_bits = {}
        ordinal = 0
        ops = []
        for line in lines:
            a = line.split()
            if not a or a[0] in ["APPEND_TO_REGISTER", "REGISTER", "INFO_TAG"]:
                continue
            op = a[0]
            qs = [int(x[1:]) for x in a[1:] if re.fullmatch("q\\d+", x)]
            cond = a[a.index("if") + 1 :] if "if" in a else []
            ops.append((op, qs, cond, a))
        i = 0
        while i < len(ops):
            op, qs, cond, a = ops[i]
            active = FULL if not cond else FULL.clone()
            assert not cond or (
                op == "CZ" and len(cond) == 1
            ), "Nonlinear measurement dependence not supported"
            if op == "CX" and (not cond):
                control = qs[0]
                targets = [qs[1]]
                i += 1
                while (
                    i < len(ops)
                    and ops[i][0] == "CX"
                    and (not ops[i][2])
                    and (ops[i][1][0] == control)
                ):
                    targets.append(ops[i][1][1])
                    i += 1
                targets = [v for v, n in Counter(targets).items() if n % 2]
                if targets:
                    tt = torch.tensor(targets, device=dev)
                    q[tt] ^= q[control][None, :]
                continue
            if op == "R":
                assert not (q[qs[0]] & active).any().item(), "Reset would discard live information"
                q[qs[0]] &= ~active
            elif op == "X":
                q[qs[0]] ^= active
            elif op == "CX":
                q[qs[1]] ^= q[qs[0]] & active
            elif op == "CCX":
                q[qs[2]] ^= q[qs[0]] & q[qs[1]] & active
            elif op == "HMR":
                b = next((x for x in a[1:] if re.fullmatch("b\\d+", x)))
                phase_bits[b] = len(phase_coefficients)
                phase_coefficients.append(q[qs[0]].clone())
                q[qs[0]] = 0
            elif op == "CZ":
                term = q[qs[0]] & q[qs[1]]
                if cond:
                    phase_coefficients[phase_bits[cond[0]]] ^= term
                else:
                    phase ^= term
            else:
                raise ValueError((op, a))
            i += 1
        delta = words[:, k * 5 : k * 5 + 5] ^ (
            words[:, (k - 1) * 5 : k * 5] if k else torch.zeros_like(words[:, :5])
        )
        expected = target ^ delta[addr.reshape(-1)].reshape(W, 63, 5).permute(2, 0, 1)
        expected_bits = torch.stack(
            [pack(expected[j] >> b & 1) for j in range(5) for b in range(8)]
        )
        assert torch.equal(q[:11], sel), (k, "address")
        assert not q[51:].any().item(), (k, "scratch")
        valid = ids < N

        def count(mask):
            return int(
                ((mask[:, None] >> torch.arange(63, device=dev) & 1).bool() & valid).sum().item()
            )

        output_mask = (
            (q[11:51] ^ expected_bits).bitwise_or(torch.zeros_like(expected_bits)).sum(0)
            if False
            else torch.zeros(W, device=dev, dtype=torch.int64)
        )
        for row in q[11:51] ^ expected_bits:
            output_mask |= row
        phase_bad = count(phase)
        for coeff in phase_coefficients:
            phase_bad += count(coeff)
        first = None
        flags = (output_mask[:, None] >> torch.arange(63, device=dev) & 1).bool() & valid
        if bool(flags.any().item()):
            flat = int(torch.nonzero(flags.flatten())[0].item())
            wi, bi = divmod(flat, 63)
            first = {
                "address": int(addr[wi, bi].item()),
                "target": [int(target[j, wi, bi].item()) for j in range(5)],
                "traversal_case": int((cases[wi, bi] % 2).item()),
                "actual": [
                    sum(((int(q[11 + j * 8 + b, wi].item()) >> bi & 1) << b for b in range(8)))
                    for j in range(5)
                ],
                "expected": [int(expected[j, wi, bi].item()) for j in range(5)],
            }
        records.append(
            {
                "batch": k,
                "value_failures": count(output_mask),
                "phase_failures": phase_bad,
                "first_value_counterexample": first,
            }
        )
        print(json.dumps(records[-1]), flush=True)
    torch.mps.synchronize()
    report = {
        "device": "mps",
        "status": (
            "FAIL" if any((r["value_failures"] or r["phase_failures"] for r in records)) else "PASS"
        ),
        "scope": "Exact selected guarded-CZ-normalized loader assembly; all1363 addresses, both reachable incoming batches, every independent HMR formal coefficient. Values, phase, address and scratch checked; no arbitrary-target claim.",
        "basis_trajectories": N * 15,
        "phase_method": "Every HMR event has its own formal variable; final phase coefficient of each must vanish. All classical gates are measurement independent; only single-bit-conditioned CZ allowed.",
        "batches": records,
        "seconds": time.monotonic() - t,
    }
    (root / "symbolic_phase_mps_report.json").write_text(json.dumps(report, indent=2))
    print(json.dumps(report), flush=True)
```

```python
def compare_inverse_mps(candidate_path, control_paths):
    import os

    os.environ["OMP_NUM_THREADS"] = "1"
    os.environ["PYTORCH_ENABLE_MPS_FALLBACK"] = "0"
    os.environ["VECLIB_MAXIMUM_THREADS"] = "1"
    import torch, time, json, re
    from pathlib import Path

    torch.set_num_threads(1)
    assert torch.backends.mps.is_available()
    N = 3133452
    MASK = (1 << 63) - 1
    W = (N + 62) // 63
    t0 = time.monotonic()
    xs = torch.arange(1, W * 63 + 1, device="mps", dtype=torch.int64).reshape(W, 63)
    weights = torch.ones(63, device="mps", dtype=torch.int64) << torch.arange(
        63, device="mps", dtype=torch.int64
    )
    inputs = torch.stack([((xs >> j & 1) * weights).sum(dim=1) for j in range(22)])
    lanes = torch.arange(W, device="mps", dtype=torch.int64)

    def simulate(path, stream):
        lines = Path(path).read_text().splitlines()
        maxq = max(
            (
                int(q)
                for l in lines
                if not l.startswith("INFO_TAG")
                for q in re.findall("\\bq(\\d+)\\b", l)
            )
        )
        q = torch.zeros((maxq + 1, W), device="mps", dtype=torch.int64)
        q[:22] = inputs
        bits = {}
        active = torch.full((W,), MASK, device="mps", dtype=torch.int64)
        stack = []
        parity = torch.zeros_like(active)
        rotations = []
        ordinal = 0

        def pred(ts):
            out = torch.full_like(active, MASK)
            for tok in ts:
                if tok in ["if", "and", "&"]:
                    continue
                neg = tok.startswith("!")
                name = tok[1:] if neg else tok
                value = bits.get(name, 0)
                out &= value ^ MASK if neg else value
            return out

        for line in lines:
            a = line.split()
            if not a:
                continue
            op = a[0]
            if op in ["APPEND_TO_REGISTER", "REGISTER", "INFO_TAG"]:
                continue
            if op == "PUSH_CONDITION":
                stack.append(active)
                active = active & pred(a[1:])
                continue
            if op == "POP_CONDITION":
                active = stack.pop()
                continue
            act = active & pred(a[a.index("if") + 1 :]) if "if" in a else active
            ids = [int(s[1:]) for s in a[1:] if re.fullmatch("q\\d+", s)]
            if op in ["BIT_STORE0", "BIT_STORE1", "BIT_INVERT"]:
                for name in a[1 : a.index("if") if "if" in a else len(a)]:
                    old = bits.get(name, 0)
                    bits[name] = (
                        old & ~act
                        if op == "BIT_STORE0"
                        else old | act if op == "BIT_STORE1" else old ^ act
                    )
            elif op == "R":
                q[ids[0]] &= ~act
            elif op == "HMR":
                ordinal += 1
                if stream == 0:
                    draw = torch.zeros_like(active)
                else:
                    draw = lanes * 6364136223846793005 + (
                        ordinal * 1442695040888963407 + stream * 9176 & MASK
                    )
                    draw ^= draw >> 17
                    draw ^= draw << 23
                    draw ^= draw >> 13
                    draw &= MASK
                name = next((s for s in a[1:] if re.fullmatch("b\\d+", s)))
                bits[name] = bits.get(name, 0) & ~act | draw & act
                parity ^= q[ids[0]] & draw & act
                q[ids[0]] &= ~act
            elif op == "X":
                q[ids[0]] ^= act
            elif op == "CX":
                q[ids[1]] ^= q[ids[0]] & act
            elif op == "CCX":
                q[ids[2]] ^= q[ids[0]] & q[ids[1]] & act
            elif op in ["CSWAP", "SWAP"]:
                if op == "CSWAP":
                    c, a, b = ids
                    mask = q[c] & act
                else:
                    a, b = ids
                    mask = act
                swap = (q[a] ^ q[b]) & mask
                q[a] ^= swap
                q[b] ^= swap
            elif op in ["Z", "CZ", "CCZ", "NEG"]:
                term = act.clone()
                for j in ids:
                    term &= q[j]
                parity ^= term
            elif op == "Z_POW":
                rotations.append((a[2], q[ids[0]].clone() & act))
            else:
                raise ValueError(line)
        assert not stack
        assert torch.equal(q[:22], inputs), "input not restored"
        assert not q[22:].any().item(), "dirty scratch"
        return (parity, rotations)

    from fractions import Fraction

    def phase_units(parity, rotations):
        out = torch.zeros((W, 63), device="mps", dtype=torch.int64)
        bit = torch.arange(63, device="mps", dtype=torch.int64)
        out += (parity[:, None] >> bit & 1) * -9223372036854775808
        for weight, values in rotations:
            amount = int(Fraction(weight) * (1 << 63)) & (1 << 64) - 1
            if amount >= 1 << 63:
                amount -= 1 << 64
            out += (values[:, None] >> bit & 1) * amount
        return out

    rows = []
    for stream in (0, 1):
        parity, rotations = simulate(candidate_path, stream)
        actual = phase_units(parity, rotations)
        for control_path in control_paths:
            parity, rotations = simulate(control_path, stream)
            expected = phase_units(parity, rotations)
            assert torch.equal(actual, expected), (control_path, stream, "phase")
            rows.append(
                dict(
                    stream=stream,
                    inputs=N,
                    control=str(control_path),
                    phase="PASS",
                    input="restored",
                    scratch="zero",
                )
            )
        del actual, parity, rotations, expected
        torch.mps.empty_cache()
    torch.mps.synchronize()
    return rows
```

```python
def audit_hmr_windows_mps(window_pairs):
    import os

    os.environ["PYTORCH_ENABLE_MPS_FALLBACK"] = "0"
    import torch

    torch.set_num_threads(1)
    assert torch.backends.mps.is_available(), "MPS required; CPU fallback forbidden"

    def run(lines, wires, states, outcomes):
        q = {wire: (states >> k & 1).to(torch.int32) for k, wire in enumerate(wires)}
        bits = {}
        phase = torch.zeros_like(states)
        measured = []
        for line in lines:
            parts = line.split()
            gate = parts[0]
            conditional = "if" in parts
            mask = bits[parts[-1]] if conditional else torch.ones_like(states)
            targets = [int(x[1:]) for x in parts[1:] if x.startswith("q")]
            if gate == "CX":
                a, b = targets
                q[b] = q[b] ^ q[a] & mask
            elif gate == "CCX":
                a, b, c = targets
                q[c] = q[c] ^ q[a] & q[b] & mask
            elif gate == "HMR":
                assert not conditional and (not measured)
                a = targets[0]
                measured.append(q[a].clone())
                phase ^= q[a] & outcomes
                q[a] = torch.zeros_like(states)
                bits[parts[2]] = outcomes
            elif gate == "CZ":
                a, b = targets
                phase ^= q[a] & q[b] & mask
            elif gate == "Z":
                phase ^= q[targets[0]] & mask
            else:
                raise AssertionError(("Outside local proof scope", line))
        assert len(measured) == 1
        return (q, phase, measured[0])

    def differences(left, right):
        value = torch.zeros_like(left[1])
        for wire in left[0]:
            value |= left[0][wire] ^ right[0][wire]
        return dict(
            value_failures=int(value.sum().item()),
            phase_failures=int((left[1] ^ right[1]).sum().item()),
            measurement_input_failures=int((left[2] ^ right[2]).sum().item()),
        )

    records = []
    for index, window in enumerate(window_pairs):
        old, new = window["before"], window["after"]
        wires = sorted(
            {int(s[1:]) for line in old + new for s in line.split() if s.startswith("q")}
        )
        assert len(wires) == 10
        states = torch.arange(1024, device="mps", dtype=torch.int32).repeat_interleave(2)
        outcomes = torch.arange(2, device="mps", dtype=torch.int32).repeat(1024)
        left, right = run(old, wires, states, outcomes), run(new, wires, states, outcomes)
        observed = differences(left, right)
        assert not any(observed.values()), (index, observed)
        records.append(dict(window=index, cases=2048, **observed))
        if index == 0:
            negative = new[:]
            assert negative[20] == "CX q14 q19"
            del negative[20]
            failures = differences(left, run(negative, wires, states, outcomes))
            assert failures == dict(
                value_failures=512, phase_failures=0, measurement_input_failures=0
            )
    return records
```

```python
def audit_loader_negative_mps(data_dir):
    # This control was selected before inspecting failures; it is not tuned.
    data_dir = Path(data_dir)
    payload = json.loads((data_dir / "angle_kmxs.json").read_text())
    lines = payload["angle_k_4"].splitlines()
    first = next(i for i, line in enumerate(lines) if line.startswith("CZ ") and " if " in line)
    assert lines[first] == "CZ q61 q10 if b0"
    del lines[first]
    payload["angle_k_4"] = "\n".join(lines) + "\n"
    negative = data_dir / "negative_first_cz"
    negative.mkdir(exist_ok=True)
    shutil.copyfile(data_dir / "submission.npz", negative / "submission.npz")
    (negative / "angle_kmxs.json").write_text(json.dumps(payload))
    run_symbolic_mps_audit(negative)
    report = json.loads((negative / "symbolic_phase_mps_report.json").read_text())
    assert sum(r["value_failures"] for r in report["batches"]) == 0
    assert sum(r["phase_failures"] for r in report["batches"]) == 6
    return report
```
