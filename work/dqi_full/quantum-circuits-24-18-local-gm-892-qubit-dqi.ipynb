## How the score works

Each task has an official verifier. It runs the circuit on many sampled inputs, checks every answer exactly, and bills a cost:

| Task | Cost | The letters |
|---|---|---|
| invsqrt | $Q \cdot C$ | $Q$ qubits; $C = N_{\mathrm{Tof}} + N_T / 4$, the mean Toffolis and T gates of one shot |
| DQI | $R \cdot Q^2 \cdot T$ | $T$ the mean Toffolis of one shot; $R \ge 1$ grows with the failed shots |
| Chemistry | $\lambda_{\mathrm{eff}} \cdot T \cdot Q$ | $\lambda_{\mathrm{eff}}$ the Hamiltonian's effective normalization; $T$ the Toffolis of one step |

A task's **ratio** is the organizers' reference cost divided by this cost. The references are 2,076,774, $2.4439 \times 10^{13}$ and 3,355,599,588. The score is the **geometric mean of the three ratios**:

$$\text{score} = \left( \frac{C^{\mathrm{ref}}_{\mathrm{invsqrt}}}{C_{\mathrm{invsqrt}}} \cdot \frac{C^{\mathrm{ref}}_{\mathrm{DQI}}}{C_{\mathrm{DQI}}} \cdot \frac{C^{\mathrm{ref}}_{\mathrm{chem}}}{C_{\mathrm{chem}}} \right)^{1/3}$$

So making any one task 1% cheaper lifts the score by about 0.33%.

The three bills reward different things. In DQI, $Q$ is squared: near 880 qubits one qubit is worth 0.23% of the cost. In invsqrt one qubit near 63 is worth 1.6%. In chemistry each of the three factors counts the same.

## invsqrt · Syed Asad Ali's 62-qubit circuit, with our normalization steps and four phase-split carry sites

**This circuit.**
- $Q = 62$. A shot runs 1,294.1651 Toffolis on average, 41 rotations and 14.5 T gates.
- The cost is $S = 98{,}743.11$, 21.0321× below the reference: 100.76 below Syed Asad Ali's circuit, at 98,843.84, on which it is built.
- Gate for gate it is that circuit, except at nine sites. The lookup, the polynomial and the 41 rotations are unchanged.

**What changed.**
- **Five normalization steps,** ported unchanged from our earlier ladder on the older public circuit: the input's normalization flags are retired one by one. The second branch splits the predicates $P_{12}$ and $P_{20}$, for $1/8$ Toffoli, and the first branch splits four of the flags it holds as XORs, for $1/4$ each. Together $1.125$ Toffolis fewer a shot.
- **Four phase-split carry sites.** At these sites the circuit writes a carry's last AND straight into its output and pays the output's phase with three T and two S gates conditioned on a phase bit, $1.5$ expected T. We branch on that bit. On phase 0 those phase lines do nothing, and they go. On phase 1 the older circuit's window for the same site takes over: it stores the AND on a clean wire and pays one CCZ on a fresh outcome, with no T. Each site costs $1/4$ Toffoli more and $3/2$ T fewer, $1/8$ of a Toffoli-equivalent less, since a T counts a quarter.
- In all, $1/8$ Toffoli and 6 T fewer a shot, so the Toffoli-equivalent cost falls by exactly $1.625$ a shot.

**How it was checked.**
- Each window is the union of the text differences of the two circuits around its site, so every line outside it is common. At build, with every other wire, classical bit and measurement outcome free, a checker shows that on phase 1 the older window gives the same state and the same sign, that on phase 0 the dropped lines change nothing, and that the classical bits that differ are overwritten before any later read.
- Three more sites of the same kind fail that check, because the carry chains moved there; they are left as they were.
- The official verifier passes at its defaults and at three fresh salts. Our replay of every outcome passes on all 3,133,452 inputs for every assignment of the 899 measurements, with exact mean Toffolis $169628799/131072$.
