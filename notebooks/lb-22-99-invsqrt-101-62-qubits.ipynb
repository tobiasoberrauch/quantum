## Submission file

The next cell writes `/kaggle/working/submission.zip` with exactly 18 verified files: one invsqrt KMB, one DQI KMB, and 16 chemistry files. It streams every input and output member while checking SHA-256, confirms the archive member set and CRCs, and records the output ZIP hash. It packages already accepted files; it does not rerun the full verifiers or submit to the competition.


```python
from pathlib import Path
from zipfile import ZipFile, ZIP_DEFLATED, ZipInfo
import hashlib
import json
import shutil

EXPECTED = {
    'dfthcsos/angle_k_0.kmb': '9c31da0c1d047e7dcac638df807d3cef96eac6f736a8838a0ae6943e9c46c74f',
    'dfthcsos/angle_k_1.kmb': '39b8d10098be097f75148639ee302249bdabe112a9278a48565eadb0cb86810b',
    'dfthcsos/angle_k_10.kmb': 'e75f6b0685d754bf76f3f4d62589bf1d07b118faeeb0ab8745833733286db2fc',
    'dfthcsos/angle_k_11.kmb': '773e54f6f0d7c3f769db433f263d541cdcff1076fb75109740907e3249996141',
    'dfthcsos/angle_k_12.kmb': 'd86ca7ddd984059ded765b953cc936388517a7ca5789676ae90910a2be89769d',
    'dfthcsos/angle_k_13.kmb': '50d82cf65116f06c0fe469ebba312f72675b277263461ab88b4d15eca947c050',
    'dfthcsos/angle_k_14.kmb': 'ae2070998e31b7e05d01183914df20b6974c1760bddf90d1adac86ea75c7bf3f',
    'dfthcsos/angle_k_2.kmb': '2e8dfdeb69c496e9e510fe99722153663760980d45c35ad9b72e79ba799dde49',
    'dfthcsos/angle_k_3.kmb': '13817cf8d36ea112e62d7863edde2193aa5907dd0605612c4f0e1e54bacc173b',
    'dfthcsos/angle_k_4.kmb': '8dce8f4a380998d818bf8a08e4f3daa62aaeb31a25d084aa57549f931a937f3d',
    'dfthcsos/angle_k_5.kmb': '42cf7d76ef0f0b36640a9e65005214fdd4d9a9bb82b8cef6e658dbbdd8417099',
    'dfthcsos/angle_k_6.kmb': '7fd7bd333b1427ecf546b0c27717ed9428e255e5165c4db14e5f95f8af45c838',
    'dfthcsos/angle_k_7.kmb': '809a775b98836ee00e8783d3f616bb8517bb3a3621978bc704d20b543a8df4ec',
    'dfthcsos/angle_k_8.kmb': '42d0deb376981d0470bebdd370ca2adf79685b4bf3e6909be262e2beced030ec',
    'dfthcsos/angle_k_9.kmb': 'f1c49e7cf8e4f3a9adaffe167d2bbe21497a52eca2c7d09f765fc5a125e304c3',
    'dfthcsos/submission.npz': 'e76b6d01229c0bdc2195efee2457a0e17a975930a4e8f097268127b6559e2584',
    'dqi.kmb': '25db915d86ab9bf2cf41670ee97b7d32495fcb12f12aa48e89c1a5924e8921ec',
    'invsqrt.kmb': '67482db6c53896348eadba126924868ed3011d7e1576a6d0abb6f7c137ae4997',
}
source = DATASET / "submission"
working = Path("/kaggle/working")
working.mkdir(parents=True, exist_ok=True)
temporary = working / "submission.tmp"
archive_path = working / "submission.zip"
try:
    with ZipFile(temporary, "w", compression=ZIP_DEFLATED, compresslevel=1) as archive:
        for name, expected_sha256 in sorted(EXPECTED.items()):
            path = source / name
            info = ZipInfo(name, date_time=(1980, 1, 1, 0, 0, 0))
            info.compress_type = ZIP_DEFLATED
            info.create_system = 3
            info.external_attr = 0o100644 << 16
            info.file_size = path.stat().st_size
            digest = hashlib.sha256()
            with path.open("rb") as source_file, archive.open(info, "w") as target_file:
                for chunk in iter(lambda: source_file.read(8 * 1024 * 1024), b""):
                    digest.update(chunk)
                    target_file.write(chunk)
            if digest.hexdigest() != expected_sha256:
                raise ValueError(f"Unexpected circuit identity: {name}")
    temporary.replace(archive_path)
except Exception:
    temporary.unlink(missing_ok=True)
    raise
with ZipFile(archive_path, "r") as archive:
    names = archive.namelist()
    if len(names) != 18 or len(set(names)) != 18 or names != sorted(EXPECTED):
        raise ValueError(f"Unexpected submission ZIP member set: {names}")
    corrupt = archive.testzip()
    if corrupt is not None:
        raise ValueError(f"Corrupt submission ZIP member: {corrupt}")
    for name, expected_sha256 in EXPECTED.items():
        digest = hashlib.sha256()
        with archive.open(name, "r") as packaged_file:
            for chunk in iter(lambda: packaged_file.read(8 * 1024 * 1024), b""):
                digest.update(chunk)
        if digest.hexdigest() != expected_sha256:
            raise ValueError(f"Packaged member identity changed: {name}")
with archive_path.open("rb") as file:
    output_sha256 = hashlib.file_digest(file, "sha256").hexdigest()
provenance = json.loads((DATASET / "portfolio-manifest.json").read_text())
provenance["packaged_output_zip_sha256"] = output_sha256
provenance["verified_input_files"] = len(EXPECTED)
(working / "submission-provenance.json").write_text(json.dumps(provenance, indent=2))
print("Ready:", archive_path)
print("All 18 submission members matched their pinned SHA-256 identities.")
print("Output ZIP SHA-256:", output_sha256)
print("This cell packages accepted artifacts; it does not run the verifiers or submit a score.")
```
