```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# 1. Quantenschaltkreis mit 2 Qubits und 2 klassischen Bits erstellen
qc = QuantumCircuit(2, 2)

# 2. Hadamard-Gatter auf Qubit 0 anwenden (Superposition)
qc.h(0)

# 3. CNOT-Gatter anwenden (Verschränkung von Qubit 0 und Qubit 1)
qc.cx(0, 1)

# 4. Beide Qubits in die klassischen Bits messen
qc.measure([0, 1], [0, 1])

# 5. Auf dem lokalen Simulator ausführen (1024 Shots)
simulator = AerSimulator()
result = simulator.run(qc, shots=1024).result()

# 6. Ergebnisse auslesen und anzeigen
counts = result.get_counts()
print(counts)

plot_histogram(counts)
plt.show()
```
