## 1. Composition and derivation

**Inverse.** Start with the exact Isabe V5 inverse and apply the two public
SomeJoint HMR windows, then require original SDK serialization to reproduce
the exact public `1511…` binary and `03252…` canonical stream. Apply Syed's
public five-to-four decode to recover our V7 canonical stream. Every edit is
an ordinary readable gate list below; no contestant generator is executed.

For a phase-free X/CX/CCX window, each output wire is a square-free Boolean
polynomial in the boundary bits: X adds 1, CX adds its control polynomial,
and CCX adds the product of its two control polynomials. Equal output
polynomials on every wire prove equal basis permutations, hence the same
linear quantum operation on arbitrary superpositions. We check all 26 wire
polynomials for Qiu's complete 82-to-95-line window. It contains 3 CCX before
and 1 after; the full credited site goes from 4 to 2 after Leo's first merge
and Qiu's further affine-conjugated merge. The resource gain is theirs.

Our conservative zero certificate on the V7 stream proves q21=0 immediately
before each selected HMR, b417 and b418, over its symbolic input and branch
histories. Qiu's window is earlier, phase-free and has the identical complete
boundary map, so those facts transfer compositionally. HMR on |0> contributes
no input-dependent phase. Each outcome is an internal, non-output bit; a
following BIT_STORE0 fixes its cleanup branch. The zero certificate proves
these selected zero facts, while the full-program comparison and native
accuracy check have the separate scopes reported below. Both HMRs stay in
place, and the official verifier, salt and sample generator are untouched.

**Current chemistry.** Replace the previous (9,143,39) factor/306–292 loaders
with Lera's (8,143,45) compact-group table and its fifteen compiled loaders,
then use SomeJoint's public W/H2_I refit. We transport the exact public data;
the unpublished compact-group/compiler/refit generator is not rebuilt.
The full original verifier recovers the loaded words and evaluates the
unchanged Hamiltonian, shifts, accuracy limits and gap floor. Q falls428→427,
T15525→15459; lambda rises47.0214→47.1847, while their product cost falls
312443243.840475→311465857.0666053. All15 changed binaries receive a new audit.

**Current DQI.** The original SDK finds exactly five changed regions between
06aa and748b; the streams agree exactly outside those regions. The
edit transports a cached product and replaces one CCZ by CZ, HMR and its
conditional CZ correction. For outcome m, erasing cached p contributes
(-1)^(m*p); the final correction contributes (-1)^(m*c*d). They cancel when
the transported starting product equals the ending product c*d, with the
specified carriers clean. Our320-case MPS kernel checks every local boundary
assignment satisfying that condition and both HMR outcomes. Removing the
new final conditional CZ yields0 value/16 phase failures. This conditional
kernel is not a proof of global product equality or whole DQI correctness.
The new full original sample has1 value error; the metric prices it rather
than imposing universal zero error. The complete receipt is below.

```python
archive_hashes = {p: sha256(p) for p in INPUT.rglob("submission.zip")}
folders, actual_sources = {}, {}
for label, pin in SOURCE_PINS.items():
    matching = [p for p, h in archive_hashes.items() if h == pin["archive"]]
    assert len(matching) == 1, (label, "Pinned ZIP missing; reject unexpected latest bytes")
    source = matching[0]
    folder = Path(tempfile.mkdtemp(prefix="exact-" + label + "-"))
    with zipfile.ZipFile(source) as archive:
        assert len(archive.namelist()) == len(set(archive.namelist())) == 18
        assert set(archive.namelist()) == set(pin["members"])
        for name, expected in pin["members"].items():
            with archive.open(name) as stream:
                assert hashlib.file_digest(stream, "sha256").hexdigest() == expected, (label, name)
            # Keep only files needed for the selected build; every member was checked above.
            needed = ((label == "isabe" and name == "invsqrt.kmb")
                      or (label == "syed" and name == "invsqrt.kmb")
                      or (label == "somejoint34" and (name.startswith("dfthcsos/") or name == "dqi.kmb")))
            if needed:
                target = folder / name
                target.parent.mkdir(parents=True, exist_ok=True)
                target.write_bytes(archive.read(name))
    folders[label] = folder
    actual_sources[label] = dict(path=str(source), archive_sha256=sha256(source), **pin)
competition = next(p.parents[2] for p in INPUT.rglob("verify_submission_invsqrt.py"))
sys.path[:0] = [str(competition / "kickmix"), str(competition / "verifiers")]
import kickmix as km
from invsqrt import verify_submission_invsqrt as original_inverse
from dfthcsos import verify as original_chemistry

display(
    pd.DataFrame(
        [dict(Source=k, Version=v["version"], ZIP=v["archive"]) for k, v in SOURCE_PINS.items()]
    )
)
```

```python
def certify_and_specialize_free_coins(raw):
    lines = raw.splitlines()
    zero, one = frozenset(), frozenset({0})
    next_atom = 100
    products = {}

    def unknown():
        nonlocal next_atom
        next_atom += 1
        return frozenset({next_atom})

    def product(a, b):
        if not a or not b:
            return zero
        if a == one:
            return b
        if b == one or a == b:
            return a
        key = frozenset((a, b))
        if key not in products:
            products[key] = unknown()
        return products[key]

    q = [unknown() if i < 22 else zero for i in range(62)]
    stack = []
    certificates = []
    selected = {"b417", "b418"}
    for line_number, line in enumerate(lines):
        tokens = line.split()
        if not tokens:
            continue
        op = tokens[0]
        if op == "PUSH_CONDITION":
            stack.append(q.copy())
            continue
        if op == "POP_CONDITION":
            before = stack.pop()
            q = [a if a == b else unknown() for a, b in zip(before, q)]
            continue
        wires = [int(t[1:]) for t in tokens[1:] if re.fullmatch(r"q\d+", t)]
        if op == "HMR" and tokens[-1] in selected:
            assert not stack and tokens == ["HMR", "q21", tokens[-1]]
            assert q[21] == zero, ("Cannot prove selected measured qubit zero", line_number, q[21])
            certificates.append(dict(line=line_number, gate=line, symbolic_value=[], condition_depth=0,
                                     proof="Conservative affine/opaque-product abstract state is identically zero for all 22-bit inputs and all branch/HMR histories, scratch initially zero."))
        old = q.copy()
        if op == "X":
            q[wires[0]] ^= one
        elif op == "CX":
            q[wires[1]] ^= q[wires[0]]
        elif op == "CCX":
            q[wires[2]] ^= product(q[wires[0]], q[wires[1]])
        elif op in ("R", "HMR"):
            q[wires[0]] = zero
        elif op == "SWAP":
            a, b = wires
            q[a], q[b] = q[b], q[a]
        elif op == "CSWAP":
            c, a, b = wires
            difference = product(q[c], q[a] ^ q[b])
            q[a] ^= difference
            q[b] ^= difference
        elif op in ("Z", "CZ", "CCZ", "NEG", "Z_POW", "REGISTER", "APPEND_TO_REGISTER", "INFO_TAG", "BIT_STORE0", "BIT_STORE1", "BIT_INVERT"):
            pass
        else:
            raise ValueError((line_number, line))
        if "if" in tokens:
            q = [a if a == b else unknown() for a, b in zip(old, q)]
    assert not stack and len(certificates) == 2
    assert not any(line.startswith("APPEND_TO_REGISTER b417 ") or line.startswith("APPEND_TO_REGISTER b418 ") for line in lines)
    output = []
    for line in lines:
        output.append(line)
        if line in ("HMR q21 b417", "HMR q21 b418"):
            output.append("BIT_STORE0 " + line.split()[-1])
    candidate = "\n".join(output) + "\n"

    return candidate, certificates
```

## 3. Reachable loaders and measurement phases

All **1,220 valid addresses**, both reachable incoming five-word batches and
all fifteen actual streams were checked: **36,600 trajectories**. All91,500
words satisfy bit7=bit6. This is the reachable forward/reverse contract
clarified by [host Tanuj Khattar](https://www.kaggle.com/competitions/google-verifiable-quantum-circuit-compilation/discussion/753092),
not an arbitrary-target XOR claim.

For every HMR event our MPS engine introduces an independent formal phase
coefficient. Every final coefficient and the constant phase vanish; address
and scratch restore, and every R acts on zero. The preregistered negative
removes the first conditional CZ in batch4, **CZ q123 q3 if b0**:0 value and
152 formal phase-coefficient failures over2,440 cases. These are not native
grading error counts and do not reuse the old292-family negative6/2726.

The older guarded-CZ reader first rejected these new bytes because its scope
required128 qubits: batches7 and13 use126. Its assertion was retained. The
existing stricter opcode-limited loader reader already supports up to128 and
agreed exactly with our independent inverse-format codec on all15 streams.
These streams need no PUSH/CZ/POP normalization. Original SDK canonical
exports subsequently matched every original stream, before full native.


```python
def normalize_loader_cz(text):
    guard, normalized = None, []
    for line in text.splitlines():
        fields = line.split()
        if fields[0] in ("REGISTER", "APPEND_TO_REGISTER"):
            assert guard is None
            normalized.append(line)
        elif fields[0] == "PUSH_CONDITION":
            assert guard is None and len(fields) == 3 and fields[1] == "if"
            assert re.fullmatch(r"b\d+", fields[2])
            guard = fields[2]
        elif fields[0] == "POP_CONDITION":
            assert guard is not None and len(fields) == 1
            guard = None
        elif guard is not None:
            assert fields[0] == "CZ" and len(fields) == 3
            normalized.append(line + " if " + guard)
        else:
            assert fields[0] in ("X", "CX", "CCX", "R", "HMR", "CZ")
            if "if" in fields:
                assert fields[0] == "CZ" and len(fields) == 5 and fields[3] == "if"
            normalized.append(line)
    assert guard is None
    return "\n".join(normalized) + "\n"


loader_texts, original_loader_texts, export_rows = {}, {}, []
export_dir = WORK / "mps_loader_inputs"
export_dir.mkdir(exist_ok=True)
shutil.copyfile(submission / "dfthcsos/submission.npz", export_dir / "submission.npz")
for k in range(15):
    name = f"angle_k_{k}"
    output = export_dir / (name + ".kmx")
    km.Circuit.from_file(str(submission / "dfthcsos" / (name + ".kmb")), format="kmb").to_file(
        str(output), format="kmx"
    )
    assert sha256(output) == EXPECTED_LOADER_CANONICAL[str(k)], k
    original_loader_texts[name] = output.read_text()
    loader_texts[name] = normalize_loader_cz(output.read_text())
    (export_dir / (name + ".normalized.kmx")).write_text(loader_texts[name])
    export_rows.append(
        dict(
            batch=k,
            CCX=sum(l.startswith("CCX ") for l in loader_texts[name].splitlines()),
            canonical_sha256=sha256(output),
        )
    )
assert [row["CCX"] for row in export_rows] == [307, 307, 307, 307, 285, 285, 285, 282, 285, 285, 285, 285, 285, 284, 285]
(export_dir / "angle_kmxs.json").write_text(json.dumps(loader_texts))
display(pd.DataFrame(export_rows))
```

### Independent Mac MPS comparisons and negative controls

The functions below are readable reproduction tools. Numerical tensors use
Mac MPS with no CPU fallback; the Kaggle CPU saved run does not invoke them.
Our completed receipts record each audit, its exact bytes and scope.

The retained inverse has the previous all3,133,452-input MPS comparisons at
two deterministic HMR streams plus forced selected coins, local4096-case HMR
windows and the exact26-wire ANF identity. This is not numerical enumeration
of every whole-program HMR history. Its fixed missingCCX negative failed
input restoration in both streams before a phase count was assigned.

The new compact427 audit and its0/152 negative have their own records; no
old40890-case proof transfers to changed binaries. The new748b cache audit
has320 conditional cases and0/16 negative failures. Global cache equality
remains INCOMPLETE. Historical06aa80/negative4 and unselected2e72400-case
audits remain explicitly historical. Old receipt NOTRUN/PENDING labels
describe their preregistration time; later original PASS markers record completion.


```python
def run_symbolic_mps_audit(data_dir, batches=tuple(range(15))):
    """MPS proof for normalized narrow reachable loaders; not arbitrary-target XOR."""
    import os

    os.environ["OMP_NUM_THREADS"] = "1"
    os.environ["PYTORCH_ENABLE_MPS_FALLBACK"] = "0"
    import torch, numpy as np, json, re, time
    from pathlib import Path
    from collections import Counter

    torch.set_num_threads(1)
    assert torch.backends.mps.is_available()
    t = time.monotonic()
    dev = "mps"
    root = Path(data_dir)
    MASK = (1 << 63) - 1
    with np.load(root / "submission.npz") as z:
        words = torch.from_numpy(z["angle_words"].copy()).to(dev, dtype=torch.int64) & 255
    A, columns = words.shape
    N = A * 2
    W = (N + 62) // 63
    ids = torch.arange(W * 63, device=dev, dtype=torch.int64).reshape(W, 63)
    addr = ids % A
    cases = ids // A
    weights = torch.ones(63, device=dev, dtype=torch.int64) << torch.arange(
        63, device=dev, dtype=torch.int64
    )

    def pack(bits):
        return (bits * weights).sum(-1)

    sel = torch.stack([pack(addr >> j & 1) for j in range(11)])
    phase_coefficients = []
    phase_bits = {}
    FULL = torch.full((W,), MASK, device=dev, dtype=torch.int64)
    payload = json.loads((root / "angle_kmxs.json").read_text())
    records = []
    for k in batches:
        text = payload["angle_k_" + str(k)]
        lines = text.splitlines()
        current = words[:, k * 5 : k * 5 + 5]
        previous = words[:, (k - 1) * 5 : k * 5] if k else torch.zeros_like(current)
        current = current[addr.reshape(-1)].reshape(W, 63, 5).permute(2, 0, 1)
        previous = previous[addr.reshape(-1)].reshape(W, 63, 5).permute(2, 0, 1)
        target = torch.where((cases % 2 == 0)[None], previous, current)
        target_bits = torch.stack([pack(target[j] >> b & 1) for j in range(5) for b in range(8)])
        q = torch.zeros((128, W), device=dev, dtype=torch.int64)
        q[:11] = sel
        q[11:51] = target_bits
        bits = {}
        phase = torch.zeros(W, device=dev, dtype=torch.int64)
        phase_coefficients = []
        phase_bits = {}
        ordinal = 0
        ops = []
        for line in lines:
            a = line.split()
            if not a or a[0] in ["APPEND_TO_REGISTER", "REGISTER", "INFO_TAG"]:
                continue
            op = a[0]
            qs = [int(x[1:]) for x in a[1:] if re.fullmatch("q\\d+", x)]
            cond = a[a.index("if") + 1 :] if "if" in a else []
            ops.append((op, qs, cond, a))
        i = 0
        while i < len(ops):
            op, qs, cond, a = ops[i]
            active = FULL if not cond else FULL.clone()
            assert not cond or (
                op == "CZ" and len(cond) == 1
            ), "Nonlinear measurement dependence not supported"
            if op == "CX" and (not cond):
                control = qs[0]
                targets = [qs[1]]
                i += 1
                while (
                    i < len(ops)
                    and ops[i][0] == "CX"
                    and (not ops[i][2])
                    and (ops[i][1][0] == control)
                ):
                    targets.append(ops[i][1][1])
                    i += 1
                targets = [v for v, n in Counter(targets).items() if n % 2]
                if targets:
                    tt = torch.tensor(targets, device=dev)
                    q[tt] ^= q[control][None, :]
                continue
            if op == "R":
                assert not (q[qs[0]] & active).any().item(), "Reset would discard live information"
                q[qs[0]] &= ~active
            elif op == "X":
                q[qs[0]] ^= active
            elif op == "CX":
                q[qs[1]] ^= q[qs[0]] & active
            elif op == "CCX":
                q[qs[2]] ^= q[qs[0]] & q[qs[1]] & active
            elif op == "HMR":
                b = next((x for x in a[1:] if re.fullmatch("b\\d+", x)))
                phase_bits[b] = len(phase_coefficients)
                phase_coefficients.append(q[qs[0]].clone())
                q[qs[0]] = 0
            elif op == "CZ":
                term = q[qs[0]] & q[qs[1]]
                if cond:
                    phase_coefficients[phase_bits[cond[0]]] ^= term
                else:
                    phase ^= term
            else:
                raise ValueError((op, a))
            i += 1
        delta = words[:, k * 5 : k * 5 + 5] ^ (
            words[:, (k - 1) * 5 : k * 5] if k else torch.zeros_like(words[:, :5])
        )
        expected = target ^ delta[addr.reshape(-1)].reshape(W, 63, 5).permute(2, 0, 1)
        expected_bits = torch.stack(
            [pack(expected[j] >> b & 1) for j in range(5) for b in range(8)]
        )
        assert torch.equal(q[:11], sel), (k, "address")
        assert not q[51:].any().item(), (k, "scratch")
        valid = ids < N

        def count(mask):
            return int(
                ((mask[:, None] >> torch.arange(63, device=dev) & 1).bool() & valid).sum().item()
            )

        output_mask = (
            (q[11:51] ^ expected_bits).bitwise_or(torch.zeros_like(expected_bits)).sum(0)
            if False
            else torch.zeros(W, device=dev, dtype=torch.int64)
        )
        for row in q[11:51] ^ expected_bits:
            output_mask |= row
        phase_bad = count(phase)
        for coeff in phase_coefficients:
            phase_bad += count(coeff)
        first = None
        flags = (output_mask[:, None] >> torch.arange(63, device=dev) & 1).bool() & valid
        if bool(flags.any().item()):
            flat = int(torch.nonzero(flags.flatten())[0].item())
            wi, bi = divmod(flat, 63)
            first = {
                "address": int(addr[wi, bi].item()),
                "target": [int(target[j, wi, bi].item()) for j in range(5)],
                "traversal_case": int((cases[wi, bi] % 2).item()),
                "actual": [
                    sum(((int(q[11 + j * 8 + b, wi].item()) >> bi & 1) << b for b in range(8)))
                    for j in range(5)
                ],
                "expected": [int(expected[j, wi, bi].item()) for j in range(5)],
            }
        records.append(
            {
                "batch": k,
                "value_failures": count(output_mask),
                "phase_failures": phase_bad,
                "first_value_counterexample": first,
            }
        )
        print(json.dumps(records[-1]), flush=True)
    torch.mps.synchronize()
    report = {
        "device": "mps",
        "status": (
            "FAIL" if any((r["value_failures"] or r["phase_failures"] for r in records)) else "PASS"
        ),
        "scope": "Exact selected guarded-CZ-normalized loader assembly; all1363 addresses, both reachable incoming batches, every independent HMR formal coefficient. Values, phase, address and scratch checked; no arbitrary-target claim.",
        "basis_trajectories": N * 15,
        "phase_method": "Every HMR event has its own formal variable; final phase coefficient of each must vanish. All classical gates are measurement independent; only single-bit-conditioned CZ allowed.",
        "batches": records,
        "seconds": time.monotonic() - t,
    }
    (root / "symbolic_phase_mps_report.json").write_text(json.dumps(report, indent=2))
    print(json.dumps(report), flush=True)
```

```python
def compare_free_coin_inverse_mps(source_path, candidate_path, negative_path):
    """Packed all-input computational-basis trajectories, including HMR phase feed-forward.
    This is a GPU comparison of two circuits, not the official quantum verifier.
    """
    import os
    os.environ['PYTORCH_ENABLE_MPS_FALLBACK'] = '0'
    os.environ['OMP_NUM_THREADS'] = '1'
    os.environ['VECLIB_MAXIMUM_THREADS'] = '1'
    import torch, time, json, re
    from pathlib import Path
    torch.set_num_threads(1)
    assert torch.backends.mps.is_available()
    N = 3133452
    MASK = (1 << 63) - 1
    W = (N + 62) // 63
    t0 = time.monotonic()
    xs = torch.arange(1, W * 63 + 1, device='mps', dtype=torch.int64).reshape(W, 63)
    weights = torch.ones(63, device='mps', dtype=torch.int64) << torch.arange(63, device='mps', dtype=torch.int64)
    inputs = torch.stack([((xs >> j & 1) * weights).sum(dim=1) for j in range(22)])
    lanes = torch.arange(W, device='mps', dtype=torch.int64)

    def simulate(path, stream, force_coins=False):
        lines = Path(path).read_text().splitlines()
        maxq = max((int(q) for l in lines if not l.startswith('INFO_TAG') for q in re.findall('\\bq(\\d+)\\b', l)))
        q = torch.zeros((maxq + 1, W), device='mps', dtype=torch.int64)
        q[:22] = inputs
        bits = {}
        active = torch.full((W,), MASK, device='mps', dtype=torch.int64)
        stack = []
        parity = torch.zeros_like(active)
        rotations = []
        ordinal = 0

        def pred(ts):
            out = torch.full_like(active, MASK)
            for tok in ts:
                if tok in ['if', 'and', '&']:
                    continue
                neg = tok.startswith('!')
                name = tok[1:] if neg else tok
                value = bits.get(name, 0)
                out &= value ^ MASK if neg else value
            return out
        for line in lines:
            a = line.split()
            if not a:
                continue
            op = a[0]
            if op in ['APPEND_TO_REGISTER', 'REGISTER', 'INFO_TAG']:
                continue
            if op == 'PUSH_CONDITION':
                stack.append(active)
                active = active & pred(a[1:])
                continue
            if op == 'POP_CONDITION':
                active = stack.pop()
                continue
            act = active & pred(a[a.index('if') + 1:]) if 'if' in a else active
            ids = [int(s[1:]) for s in a[1:] if re.fullmatch('q\\d+', s)]
            if op in ['BIT_STORE0', 'BIT_STORE1', 'BIT_INVERT']:
                for name in a[1:a.index('if') if 'if' in a else len(a)]:
                    old = bits.get(name, 0)
                    bits[name] = old & ~act if op == 'BIT_STORE0' else old | act if op == 'BIT_STORE1' else old ^ act
            elif op == 'R':
                q[ids[0]] &= ~act
            elif op == 'HMR':
                ordinal += 1
                if stream == 0:
                    draw = torch.zeros_like(active)
                else:
                    draw = lanes * 6364136223846793005 + (ordinal * 1442695040888963407 + stream * 9176 & MASK)
                    draw ^= draw >> 17
                    draw ^= draw << 23
                    draw ^= draw >> 13
                    draw &= MASK
                name = next((s for s in a[1:] if re.fullmatch('b\\d+', s)))
                if force_coins and name in ('b417', 'b418'):
                    draw = torch.full_like(active, MASK)
                bits[name] = bits.get(name, 0) & ~act | draw & act
                parity ^= q[ids[0]] & draw & act
                q[ids[0]] &= ~act
            elif op == 'X':
                q[ids[0]] ^= act
            elif op == 'CX':
                q[ids[1]] ^= q[ids[0]] & act
            elif op == 'CCX':
                q[ids[2]] ^= q[ids[0]] & q[ids[1]] & act
            elif op in ['CSWAP', 'SWAP']:
                if op == 'CSWAP':
                    c, a, b = ids
                    mask = q[c] & act
                else:
                    a, b = ids
                    mask = act
                swap = (q[a] ^ q[b]) & mask
                q[a] ^= swap
                q[b] ^= swap
            elif op in ['Z', 'CZ', 'CCZ', 'NEG']:
                term = act.clone()
                for j in ids:
                    term &= q[j]
                parity ^= term
            elif op == 'Z_POW':
                rotations.append((a[2], q[ids[0]].clone() & act))
            else:
                raise ValueError(line)
        assert not stack
        assert torch.equal(q[:22], inputs), 'input not restored'
        assert not q[22:].any().item(), 'dirty scratch'
        return (parity, rotations)
    from fractions import Fraction

    def phase_units(parity, rotations):
        out = torch.zeros((W, 63), device='mps', dtype=torch.int64)
        bit = torch.arange(63, device='mps', dtype=torch.int64)
        out += (parity[:, None] >> bit & 1) * -9223372036854775808
        for weight, values in rotations:
            amount = int(Fraction(weight) * (1 << 63)) & (1 << 64) - 1
            if amount >= 1 << 63:
                amount -= 1 << 64
            out += (values[:, None] >> bit & 1) * amount
        return out

    results = []
    for stream, force in [(0, False), (1, False), (1, True)]:
        a, b = simulate(source_path, stream, force)
        expected = phase_units(a, b)
        a, b = simulate(candidate_path, stream, force)
        actual = phase_units(a, b)
        assert torch.equal(expected, actual), "Exact total phase changed"
        results.append(dict(stream=stream, forced_selected_outcomes_1=force, inputs=N, phase="PASS", input="restored", scratch="zero"))
        if force:
            a, b = simulate(negative_path, stream, True)
            failures = int((expected != phase_units(a, b)).reshape(-1)[:N].sum().item())
            assert failures == N
        del expected, actual, a, b
        torch.mps.empty_cache()
    torch.mps.synchronize()
    return results


def compare_quadratic_inverse_mps(source_path, candidate_path, negative_path):
    import os
    os.environ['PYTORCH_ENABLE_MPS_FALLBACK'] = '0'
    os.environ['OMP_NUM_THREADS'] = '1'
    os.environ['VECLIB_MAXIMUM_THREADS'] = '1'
    import torch, time, json, re
    from pathlib import Path
    torch.set_num_threads(1)
    assert torch.backends.mps.is_available()
    N = 3133452
    MASK = (1 << 63) - 1
    W = (N + 62) // 63
    t0 = time.monotonic()
    xs = torch.arange(1, W * 63 + 1, device='mps', dtype=torch.int64).reshape(W, 63)
    weights = torch.ones(63, device='mps', dtype=torch.int64) << torch.arange(63, device='mps', dtype=torch.int64)
    inputs = torch.stack([((xs >> j & 1) * weights).sum(dim=1) for j in range(22)])
    lanes = torch.arange(W, device='mps', dtype=torch.int64)

    def simulate(path, stream, force_coins=False):
        lines = Path(path).read_text().splitlines()
        maxq = max((int(q) for l in lines if not l.startswith('INFO_TAG') for q in re.findall('\\bq(\\d+)\\b', l)))
        q = torch.zeros((maxq + 1, W), device='mps', dtype=torch.int64)
        q[:22] = inputs
        bits = {}
        active = torch.full((W,), MASK, device='mps', dtype=torch.int64)
        stack = []
        parity = torch.zeros_like(active)
        rotations = []
        ordinal = 0

        def pred(ts):
            out = torch.full_like(active, MASK)
            for tok in ts:
                if tok in ['if', 'and', '&']:
                    continue
                neg = tok.startswith('!')
                name = tok[1:] if neg else tok
                value = bits.get(name, 0)
                out &= value ^ MASK if neg else value
            return out
        for line in lines:
            a = line.split()
            if not a:
                continue
            op = a[0]
            if op in ['APPEND_TO_REGISTER', 'REGISTER', 'INFO_TAG']:
                continue
            if op == 'PUSH_CONDITION':
                stack.append(active)
                active = active & pred(a[1:])
                continue
            if op == 'POP_CONDITION':
                active = stack.pop()
                continue
            act = active & pred(a[a.index('if') + 1:]) if 'if' in a else active
            ids = [int(s[1:]) for s in a[1:] if re.fullmatch('q\\d+', s)]
            if op in ['BIT_STORE0', 'BIT_STORE1', 'BIT_INVERT']:
                for name in a[1:a.index('if') if 'if' in a else len(a)]:
                    old = bits.get(name, 0)
                    bits[name] = old & ~act if op == 'BIT_STORE0' else old | act if op == 'BIT_STORE1' else old ^ act
            elif op == 'R':
                q[ids[0]] &= ~act
            elif op == 'HMR':
                ordinal += 1
                if stream == 0:
                    draw = torch.zeros_like(active)
                else:
                    draw = lanes * 6364136223846793005 + (ordinal * 1442695040888963407 + stream * 9176 & MASK)
                    draw ^= draw >> 17
                    draw ^= draw << 23
                    draw ^= draw >> 13
                    draw &= MASK
                name = next((s for s in a[1:] if re.fullmatch('b\\d+', s)))
                if force_coins and name in ('b417', 'b418'):
                    draw = torch.full_like(active, MASK)
                bits[name] = bits.get(name, 0) & ~act | draw & act
                parity ^= q[ids[0]] & draw & act
                q[ids[0]] &= ~act
            elif op == 'X':
                q[ids[0]] ^= act
            elif op == 'CX':
                q[ids[1]] ^= q[ids[0]] & act
            elif op == 'CCX':
                q[ids[2]] ^= q[ids[0]] & q[ids[1]] & act
            elif op in ['CSWAP', 'SWAP']:
                if op == 'CSWAP':
                    c, a, b = ids
                    mask = q[c] & act
                else:
                    a, b = ids
                    mask = act
                swap = (q[a] ^ q[b]) & mask
                q[a] ^= swap
                q[b] ^= swap
            elif op in ['Z', 'CZ', 'CCZ', 'NEG']:
                term = act.clone()
                for j in ids:
                    term &= q[j]
                parity ^= term
            elif op == 'Z_POW':
                rotations.append((a[2], q[ids[0]].clone() & act))
            else:
                raise ValueError(line)
        assert not stack
        assert torch.equal(q[:22], inputs), 'input not restored'
        assert not q[22:].any().item(), 'dirty scratch'
        return (parity, rotations)
    from fractions import Fraction

    def phase_units(parity, rotations):
        out = torch.zeros((W, 63), device='mps', dtype=torch.int64)
        bit = torch.arange(63, device='mps', dtype=torch.int64)
        out += (parity[:, None] >> bit & 1) * -9223372036854775808
        for weight, values in rotations:
            amount = int(Fraction(weight) * (1 << 63)) & (1 << 64) - 1
            if amount >= 1 << 63:
                amount -= 1 << 64
            out += (values[:, None] >> bit & 1) * amount
        return out

    rows, negatives = [], []
    for stream, force in [(0, False), (1, False), (1, True)]:
        p, rotations = simulate(source_path, stream, force)
        expected = phase_units(p, rotations)
        p, rotations = simulate(candidate_path, stream, force)
        actual = phase_units(p, rotations)
        assert torch.equal(expected, actual), "Exact64bit phase changed"
        rows.append(dict(inputs=N, stream=stream, forced_selected_coins_1=force,
                         phase="PASS", input="restored", scratch="zero"))
        if not force:
            try:
                simulate(negative_path, stream, force)
            except AssertionError as error:
                assert str(error) == "input not restored"
                negatives.append(dict(stream=stream, status="EXPECTED_FAIL",
                                      observed_failure=str(error), phase_count=None))
            else:
                raise AssertionError("Predeclared missing-CCX negative did not fail restoration")
        del expected, actual, p, rotations
        torch.mps.empty_cache()
    torch.mps.synchronize()
    return dict(results=rows, negatives=negatives, seconds=time.monotonic()-t0)
```

```python
def audit_hmr_windows_mps(window_pairs):
    import os

    os.environ["PYTORCH_ENABLE_MPS_FALLBACK"] = "0"
    import torch

    torch.set_num_threads(1)
    assert torch.backends.mps.is_available(), "MPS required; CPU fallback forbidden"

    def run(lines, wires, states, outcomes):
        q = {wire: (states >> k & 1).to(torch.int32) for k, wire in enumerate(wires)}
        bits = {}
        phase = torch.zeros_like(states)
        measured = []
        for line in lines:
            parts = line.split()
            gate = parts[0]
            conditional = "if" in parts
            mask = bits[parts[-1]] if conditional else torch.ones_like(states)
            targets = [int(x[1:]) for x in parts[1:] if x.startswith("q")]
            if gate == "CX":
                a, b = targets
                q[b] = q[b] ^ q[a] & mask
            elif gate == "CCX":
                a, b, c = targets
                q[c] = q[c] ^ q[a] & q[b] & mask
            elif gate == "HMR":
                assert not conditional and (not measured)
                a = targets[0]
                measured.append(q[a].clone())
                phase ^= q[a] & outcomes
                q[a] = torch.zeros_like(states)
                bits[parts[2]] = outcomes
            elif gate == "CZ":
                a, b = targets
                phase ^= q[a] & q[b] & mask
            elif gate == "Z":
                phase ^= q[targets[0]] & mask
            else:
                raise AssertionError(("Outside local proof scope", line))
        assert len(measured) == 1
        return (q, phase, measured[0])

    def differences(left, right):
        value = torch.zeros_like(left[1])
        for wire in left[0]:
            value |= left[0][wire] ^ right[0][wire]
        return dict(
            value_failures=int(value.sum().item()),
            phase_failures=int((left[1] ^ right[1]).sum().item()),
            measurement_input_failures=int((left[2] ^ right[2]).sum().item()),
        )

    records = []
    for index, window in enumerate(window_pairs):
        old, new = window["before"], window["after"]
        wires = sorted(
            {int(s[1:]) for line in old + new for s in line.split() if s.startswith("q")}
        )
        assert len(wires) == 10
        states = torch.arange(1024, device="mps", dtype=torch.int32).repeat_interleave(2)
        outcomes = torch.arange(2, device="mps", dtype=torch.int32).repeat(1024)
        left, right = run(old, wires, states, outcomes), run(new, wires, states, outcomes)
        observed = differences(left, right)
        assert not any(observed.values()), (index, observed)
        records.append(dict(window=index, cases=2048, **observed))
        if index == 0:
            negative = new[:]
            assert negative[20] == "CX q14 q19"
            del negative[20]
            failures = differences(left, run(negative, wires, states, outcomes))
            assert failures == dict(
                value_failures=512, phase_failures=0, measurement_input_failures=0
            )
    return records
```

```python
def audit_loader_negative_mps(data_dir):
    # This control was selected before inspecting failures; it is not tuned.
    data_dir = Path(data_dir)
    payload = json.loads((data_dir / "angle_kmxs.json").read_text())
    lines = payload["angle_k_4"].splitlines()
    first = next(i for i, line in enumerate(lines) if line.startswith("CZ ") and " if " in line)
    assert lines[first] == "CZ q123 q3 if b0"
    del lines[first]
    payload["angle_k_4"] = "\n".join(lines) + "\n"
    negative = data_dir / "negative_first_cz"
    negative.mkdir(exist_ok=True)
    shutil.copyfile(data_dir / "submission.npz", negative / "submission.npz")
    (negative / "angle_kmxs.json").write_text(json.dumps(payload))
    run_symbolic_mps_audit(negative, batches=(4,))
    report = json.loads((negative / "symbolic_phase_mps_report.json").read_text())
    assert sum(r["value_failures"] for r in report["batches"]) == 0
    assert sum(r["phase_failures"] for r in report["batches"]) == 152
    return report
```

```python
def audit_historical_unselected_dqi_kernels_mps():
    import os
    os.environ['PYTORCH_ENABLE_MPS_FALLBACK'] = '0'
    """Conditional exhaustive MPS audits of both exact new SDK-derived DQI cache kernels."""
    from pathlib import Path
    import json
    import torch

    torch.set_num_threads(1)
    assert torch.backends.mps.is_available(), "MPS required; no CPU fallback"
    delta = DQI_SDK_DELTA
    assert delta["old_binary_sha256"] == "06aa8ca205096f5305d89318a3f335516686726377950143ce232e4beb3c7374"
    assert delta["new_binary_sha256"] == "2e7207b0fab0be9e2f18481f1360fbcced867b42b2d0ddc6b1a44fc34cd065e4"
    ds = delta["deltas"]
    assert len(ds) == 10 and ds[0]["old"] == ["CCX q849 q882 q881"] and ds[4]["old"] == ["CCZ q862 q878 q859"]
    assert ds[5]["old"] == ["CCX q879 q890 q881"] and ds[9]["old"] == ["CCZ q844 q852 q847"]
    families = [dict(names=["q849","q882","q862","q878","q881","q859"],scratch=["q841","q840","q855"],
                     old=ds[0]["old"]+ds[4]["old"],new=sum([x["new"] for x in ds[:5]],[]),measurement="b101623",affine=True),
                dict(names=["q879","q890","q851","q852","q881","q847","q842","q853"],scratch=["q855","q844","q848","q877"],
                     old=ds[5]["old"]+ds[7]["old"]+ds[9]["old"],new=sum([x["new"] for x in ds[5:]],[]),measurement="b101622",affine=False)]
    rows=[]
    for index, family in enumerate(families):
        n = len(family["names"])
        cases = torch.arange(1 << (n+1),device="mps",dtype=torch.int64)
        a,b,c,d = [(cases >> j)&1 for j in range(4)]
        left,right = ((a&b)^b,(c&d)^d) if family["affine"] else (a&b,c&d)
        cases = cases[left == right]
        def simulate(lines):
            q={name:((cases>>j)&1).clone() for j,name in enumerate(family["names"])}
            q.update({name:torch.zeros_like(cases) for name in family["scratch"]})
            m=(cases>>n)&1
            phase=torch.zeros_like(cases)
            for line in lines:
                op,*tokens=line.split(); wires=[x for x in tokens if x.startswith("q")]
                if op=="CCX":q[wires[2]]^=q[wires[0]]&q[wires[1]]
                elif op=="CX":q[wires[1]]^=q[wires[0]]
                elif op=="SWAP":q[wires[0]],q[wires[1]]=q[wires[1]],q[wires[0]]
                elif op in ("Z","CZ","CCZ"):
                    term=torch.ones_like(cases)
                    for wire in wires:term&=q[wire]
                    if "if" in tokens:
                        assert tokens[-1]==family["measurement"]
                        term&=m
                    phase^=term
                elif op=="HMR":
                    assert tokens[-1]==family["measurement"]
                    phase^=q[wires[0]]&m;q[wires[0]]=torch.zeros_like(cases)
                else:raise ValueError(line)
            return torch.stack([q[name] for name in family["names"]+family["scratch"]]),phase
        expected,expected_phase=simulate(family["old"])
        actual,actual_phase=simulate(family["new"])
        assert torch.equal(expected,actual) and torch.equal(expected_phase,actual_phase)
        row=dict(kernel=index,cases=int(cases.numel()),values="PASS",phase="PASS",old=family["old"],new=family["new"],
                 preconditions="Clean carrier/destinations at the stated transfer boundaries; abstract intervening common stream. " +
                 ("start(a*b XOR b) equals end(c*d XOR d)." if family["affine"] else "start(a*b) equals end(c*d); common CX q851 q844 loads the final control, included explicitly."))
        if index==0:
            removed="CZ q862 q878 if b101623"
            assert family["new"].count(removed)==1
            broken,broken_phase=simulate([x for x in family["new"] if x!=removed])
            counts=dict(value_failures=int((expected!=broken).any(0).sum().item()),phase_failures=int((expected_phase!=broken_phase).sum().item()))
            assert counts["value_failures"]==0 and counts["phase_failures"]>0
            row["negative_control"]=dict(remove=removed,expected="FAIL",**counts)
        rows.append(row)
    report=dict(status="PASS",device="mps",old_binary_sha256=delta["old_binary_sha256"],new_binary_sha256=delta["new_binary_sha256"],
                original_sdk_canonical_sha256=delta["new_canonical_sha256"],kernels=rows,cases=sum(x["cases"] for x in rows),
                exact_sdk_delta_count=10,all_other_sdk_lines_identical=True,
                scope="Exhaustive local coupled-boundary phase/value kernels only. Carrier/destination cleanliness and endpoint-function equality are stated preconditions, not a whole DQI equivalence proof. Global symbolic equality not established. Inherited approximate decoder and repetition penalty retained; separate original full102400 required.")
    return report
```

```python
def audit_selected_second_cache_mps():
    import os
    os.environ["PYTORCH_ENABLE_MPS_FALLBACK"]="0"
    import torch
    torch.set_num_threads(1)
    assert torch.backends.mps.is_available(), "Mac MPS required; no CPU fallback"
    receipt=SELECTED_DQI_MPS_RECEIPT
    names=receipt["preregistration"]["boundary_wires"]
    scratch=receipt["preregistration"]["scratch"]
    old,new=receipt["kernels"][0]["old"],receipt["kernels"][0]["new"]
    n=len(names)
    cases=torch.arange(1<<(n+1),device="mps",dtype=torch.int64)
    a,b,c,d=[(cases>>j)&1 for j in range(4)]
    cases=cases[(a&b)==(c&d)]
    def simulate(lines):
        q={name:((cases>>j)&1).clone() for j,name in enumerate(names)}
        q.update({name:torch.zeros_like(cases) for name in scratch})
        m=(cases>>n)&1;phase=torch.zeros_like(cases)
        for line in lines:
            op,*tokens=line.split();wires=[x for x in tokens if x.startswith("q")]
            if op=="CCX":q[wires[2]]^=q[wires[0]]&q[wires[1]]
            elif op=="CX":q[wires[1]]^=q[wires[0]]
            elif op=="SWAP":q[wires[0]],q[wires[1]]=q[wires[1]],q[wires[0]]
            elif op in ("Z","CZ","CCZ"):
                term=torch.ones_like(cases)
                for wire in wires:term&=q[wire]
                if "if" in tokens:assert tokens[-1]=="b101622";term&=m
                phase^=term
            elif op=="HMR":
                assert tokens[-1]=="b101622";phase^=q[wires[0]]&m;q[wires[0]]=torch.zeros_like(cases)
            else:raise ValueError(line)
        return torch.stack([q[name] for name in names+scratch]),phase
    expected,phase=simulate(old);actual,actual_phase=simulate(new)
    assert torch.equal(expected,actual) and torch.equal(phase,actual_phase)
    removed="CZ q844 q852 if b101622"
    broken,broken_phase=simulate([line for line in new if line!=removed])
    failures=dict(value_failures=int((expected!=broken).any(0).sum().item()),phase_failures=int((phase!=broken_phase).sum().item()))
    assert failures==dict(value_failures=0,phase_failures=16)
    return dict(cases=int(cases.numel()),negative=failures,scope=receipt["scope"])
```
