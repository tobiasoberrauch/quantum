## Replace 48 controlled swaps before readout

For a controlled swap on basis bits `a,b` with control `c`, let `r,s` be the following X-basis measurement outcomes. Set `d = a XOR b` with one CX. Measure `b` with `u = r XOR s`, apply `CZ(c,d)` when `u=1`, then measure `d` with `v=r`. The output bits are `(r,s)=(v,u XOR v)`.

Both forms contribute the phase exponent `r*d XOR (r XOR s)*b XOR (r XOR s)*c*d`. The outcome map is invertible. Each of the four verified windows contains 12 disjoint pairs and preserves the common control. Applying these windows to the Maoxin DQI produces the exact SHA-256 of Özer V70 and removes 48 Toffoli-like gates.

The paired MPS run checked the full ordered streams on 2,048 trajectories, including all quantum values, classical values and phase. Disabling the first new phase correction produced one phase mismatch. The separate full original verifier used its preregistered inputs and full repetition factor.

```python
PINS = {'dfthcsos/angle_k_0.kmb': 'd3aab7479fe3b8510e3af3797a14ec8d9bfa226d7aa7717ae4e93f496d2d9538',
 'dfthcsos/angle_k_1.kmb': '4595e5dcc982295db2520c8497219e1768757edf3aa9e706fa44b1716f996ed2',
 'dfthcsos/angle_k_10.kmb': 'f28f2ead2b7eb3d45aadf4e0bb9f9399c7d8f016247d7020ae39dbd9c062cace',
 'dfthcsos/angle_k_11.kmb': '7b9094ddc215c6c7db223fe319e6d2a6f67c1669f412e4d3281c280723df5b81',
 'dfthcsos/angle_k_12.kmb': 'd49dfbfdb8cff3c3d680e4ef65a3d1a01db7a2fc666f49a522d7f58f50305e78',
 'dfthcsos/angle_k_13.kmb': 'cb7e48a7029323a6ec11c0f6dede5f5e0443f67da92e0e6554ff9120b8279efd',
 'dfthcsos/angle_k_14.kmb': '07fcdcb75d3a9e7f2573e057376ec20b7a56bf9a53fe7d7f00fb351614c2003f',
 'dfthcsos/angle_k_2.kmb': '77dec28c5ebffe723fb58717a31c017d66c44cbec220bc3cc5a15a2fff051aa3',
 'dfthcsos/angle_k_3.kmb': '8e137a014a1325450eaf9ec15bc824475b757b3a05bbc542eed9cccfd6b5ec53',
 'dfthcsos/angle_k_4.kmb': 'd9f2a0d739142a1d20789352c617cccc702a8f0b4b17f91971e4e4ef3e6f6c3f',
 'dfthcsos/angle_k_5.kmb': 'bd0886bb6c3d5c04f5e1fd29dbd2562d1c7e5b1a7406bc1abde6819093a0880e',
 'dfthcsos/angle_k_6.kmb': '12c0c6f92d708ff7d852f2c6f36a40ecf4369249fa1f0ab4527b08e0152e7558',
 'dfthcsos/angle_k_7.kmb': '93dc9742392a0dd6e754aebbbea1519014292a56b6908dfce5ae417f2cde73d6',
 'dfthcsos/angle_k_8.kmb': 'd43bd96b88e6ef38c7ea9e697856646af4259889b75d1bc67a94a1768670b5a6',
 'dfthcsos/angle_k_9.kmb': 'a5960a572c39a3996b3919eb8bc18066b8aa78d3fc83bc7435d7e5550de3a2a6',
 'dfthcsos/submission.npz': 'd994cd720a6f4718560e5dfe16a52c3fe0a2a7126122e410d70f7b876d300b4b',
 'dqi.kmb': '4e9b50b1ee34b3eaf2b7f394873c2dd3dad4cddbdd1f12a348b4323b5eb41771',
 'invsqrt.kmb': '034c7c2a54d51e13aff6e62ec4c14997d3968b6c11122146c57290b1a1b8c7e2'}

ORIGINAL_PINS = {'verifiers/dfthcsos/__init__.py': '81915e324cc0cade67f52a54b7a51991fd96abaea94d1f8a4d423dae20e57348',
 'verifiers/dfthcsos/example_submission/angle_k_reference_circuit.py': '8daccdca69f2d718b0176cd9281803df84a988ea5ac6c167fbb02e461ac2b371',
 'verifiers/dfthcsos/scoreinfra/__init__.py': 'be90e0d503d5c20dcac0102af638b9a4ccad0341daa5baa5d49216a259e4516f',
 'verifiers/dfthcsos/scoreinfra/circuits.py': 'fc027b52fdcebe14404f403a089122485d097ed65d03e6d989623d620b1ab68c',
 'verifiers/dfthcsos/scoreinfra/costs_prepare.py': '7f1e190c42fc5325f3e6689fbbef51f95b0cb12f1c828cda92b6c088e72206da',
 'verifiers/dfthcsos/scoreinfra/dfthc_math.py': 'ae82898d47a7a86b96378fb906e0f8a9f46003065cf19c7c09952a123b5b5af8',
 'verifiers/dfthcsos/scoreinfra/reference.py': 'a2d5a63191d655729e29f0fde1e85fe4630bf896a91a63bb894d8d7daf71ed39',
 'verifiers/dfthcsos/verify.py': '0775be7a528922367caebe9e4d051d6199811f1aefeaa4b793eaaa4621b8ec9d',
 'verifiers/dqi/__init__.py': '01c870aa9763f4e4357fc6654bff1d3296cec995bd0943f87b4a48946ada5b40',
 'verifiers/dqi/dqi_reference_circuit.py': '2b8e1b80a806157c333a7e308eaf2499ee09de499efbc05875202c327959959e',
 'verifiers/dqi/verify_submission_dqi.py': 'f14274bc0943fb9769ddb01131be6063191a15009fe1caf1d4bdd9fd25194e65',
 'verifiers/invsqrt/__init__.py': 'e9a6ecf253c343528578be5fa92eb820ccade63333aebb784552fb4ae75dbc0b',
 'verifiers/invsqrt/nr_reference_circuit.py': 'bb90330dcf774f7b502de471293facba67f27952cc8808fe4a9970a9b1322e23',
 'verifiers/invsqrt/verify_submission_invsqrt.py': 'ddef520dec6b68d1cf0a6f328ddd36d79eb84570b9192cf3f7acaf073f06733a'}

SOURCES = {'maoxin': {'pins': {'dfthcsos/angle_k_0.kmb': 'd3aab7479fe3b8510e3af3797a14ec8d9bfa226d7aa7717ae4e93f496d2d9538',
                     'dfthcsos/angle_k_1.kmb': '4595e5dcc982295db2520c8497219e1768757edf3aa9e706fa44b1716f996ed2',
                     'dfthcsos/angle_k_10.kmb': 'f28f2ead2b7eb3d45aadf4e0bb9f9399c7d8f016247d7020ae39dbd9c062cace',
                     'dfthcsos/angle_k_11.kmb': '7b9094ddc215c6c7db223fe319e6d2a6f67c1669f412e4d3281c280723df5b81',
                     'dfthcsos/angle_k_12.kmb': 'd49dfbfdb8cff3c3d680e4ef65a3d1a01db7a2fc666f49a522d7f58f50305e78',
                     'dfthcsos/angle_k_13.kmb': 'cb7e48a7029323a6ec11c0f6dede5f5e0443f67da92e0e6554ff9120b8279efd',
                     'dfthcsos/angle_k_14.kmb': '07fcdcb75d3a9e7f2573e057376ec20b7a56bf9a53fe7d7f00fb351614c2003f',
                     'dfthcsos/angle_k_2.kmb': '77dec28c5ebffe723fb58717a31c017d66c44cbec220bc3cc5a15a2fff051aa3',
                     'dfthcsos/angle_k_3.kmb': '8e137a014a1325450eaf9ec15bc824475b757b3a05bbc542eed9cccfd6b5ec53',
                     'dfthcsos/angle_k_4.kmb': 'd9f2a0d739142a1d20789352c617cccc702a8f0b4b17f91971e4e4ef3e6f6c3f',
                     'dfthcsos/angle_k_5.kmb': 'bd0886bb6c3d5c04f5e1fd29dbd2562d1c7e5b1a7406bc1abde6819093a0880e',
                     'dfthcsos/angle_k_6.kmb': '12c0c6f92d708ff7d852f2c6f36a40ecf4369249fa1f0ab4527b08e0152e7558',
                     'dfthcsos/angle_k_7.kmb': '93dc9742392a0dd6e754aebbbea1519014292a56b6908dfce5ae417f2cde73d6',
                     'dfthcsos/angle_k_8.kmb': 'd43bd96b88e6ef38c7ea9e697856646af4259889b75d1bc67a94a1768670b5a6',
                     'dfthcsos/angle_k_9.kmb': 'a5960a572c39a3996b3919eb8bc18066b8aa78d3fc83bc7435d7e5550de3a2a6',
                     'dfthcsos/submission.npz': 'd994cd720a6f4718560e5dfe16a52c3fe0a2a7126122e410d70f7b876d300b4b',
                     'dqi.kmb': '4432ec5fd63d7d70aab92d32f719d47539d0ff66ae0cb0b964c914e01fece431',
                     'invsqrt.kmb': '034c7c2a54d51e13aff6e62ec4c14997d3968b6c11122146c57290b1a1b8c7e2'},
            'sha256': '55eec7537872f663400ab2a96ceb651ac44529445c27f6e537eea5a57c138c23',
            'version': 367139607}}

TRANSPORT_PLAN = {'baseline_header': 'd750c7d5c329d326e3cc9f6834f2b8bf020000007c030000430d030045100000b4637903000000007909060000000000eb0b3f03000000004b8e060000000000f58c2e000000000075863c0000000000000000000000000024450000000000007ed0030000000000',
 'baseline_sha256': '4432ec5fd63d7d70aab92d32f719d47539d0ff66ae0cb0b964c914e01fece431',
 'candidate_header': 'd750c7d5c329d326e3cc9f6834f2b8bf020000007c030000430d030045100000146479030000000049090600000000004b0c3f03000000004b8e060000000000258d2e0000000000d5863c0000000000000000000000000024450000000000007ed0030000000000',
 'candidate_sha256': '4e9b50b1ee34b3eaf2b7f394873c2dd3dad4cddbdd1f12a348b4323b5eb41771',
 'packets': {'1': {'new_length': 58287124,
                   'old_length': 58287028,
                   'patches': [{'after': '2150a050c22150a050c22150a050c22150a050c22150a050c22150a050c22150a050c22150a050c22150a050c22150a050c22150a050c22150a050c2',
                                'before': '323232323232323232323232505050505050505050505050505050505050505050505050',
                                'offset': 16083438},
                               {'after': '2150a050c22150a050c22150a050c22150a050c22150a050c22150a050c22150a050c22150a050c22150a050c22150a050c22150a050c22150a050c2',
                                'before': '323232323232323232323232505050505050505050505050505050505050505050505050',
                                'offset': 16233702},
                               {'after': '2150a050c22150a050c22150a050c22150a050c22150a050c22150a050c22150a050c22150a050c22150a050c22150a050c22150a050c22150a050c2',
                                'before': '323232323232323232323232505050505050505050505050505050505050505050505050',
                                'offset': 42100238},
                               {'after': '2150a050c22150a050c22150a050c22150a050c22150a050c22150a050c22150a050c22150a050c22150a050c22150a050c22150a050c22150a050c2',
                                'before': '323232323232323232323232505050505050505050505050505050505050505050505050',
                                'offset': 42250502}]},
             '10': {'new_length': 0, 'old_length': 0, 'patches': []},
             '11': {'new_length': 391590, 'old_length': 391590, 'patches': []},
             '2': {'new_length': 1582372,
                   'old_length': 1582564,
                   'patches': [{'after': '',
                                'before': '120000001200000012000000120000001200000012000000120000001200000012000000120000001200000012000000',
                                'offset': 575800},
                               {'after': '',
                                'before': '120000001200000012000000120000001200000012000000120000001200000012000000120000001200000012000000',
                                'offset': 578752},
                               {'after': '',
                                'before': '120000001200000012000000120000001200000012000000120000001200000012000000120000001200000012000000',
                                'offset': 814136},
                               {'after': '',
                                'before': '120000001200000012000000120000001200000012000000120000001200000012000000120000001200000012000000',
                                'offset': 817088}]},
             '3': {'new_length': 1582372,
                   'old_length': 1582564,
                   'patches': [{'after': '',
                                'before': '590300005a030000600300007803000050030000580300007b030000740300001400000054030000720300004d030000',
                                'offset': 575800},
                               {'after': '',
                                'before': '56030000550300007303000052030000760300005d030000530300004c0300000f0000005f0300006f03000016000000',
                                'offset': 578752},
                               {'after': '',
                                'before': '5a030000600300007803000050030000580300007b030000740300001400000054030000720300004d03000059030000',
                                'offset': 814136},
                               {'after': '',
                                'before': '0e0000006d03000057030000150000006c0300005e030000110000006e030000100000000c0000001300000048030000',
                                'offset': 817088}]},
             '4': {'new_length': 1582372,
                   'old_length': 1582564,
                   'patches': [{'after': '',
                                'before': '160000006f0300005f0300000f0000004c030000530300005d0300007603000052030000730300005503000056030000',
                                'offset': 575800},
                               {'after': '',
                                'before': '4d030000720300005403000014000000740300007b030000580300005003000078030000600300005a03000059030000',
                                'offset': 578752},
                               {'after': '',
                                'before': '48030000130000000c000000100000006e030000110000005e0300006c03000015000000570300006d0300000e000000',
                                'offset': 814136},
                               {'after': '',
                                'before': '590300004d030000720300005403000014000000740300007b030000580300005003000078030000600300005a030000',
                                'offset': 817088}]},
             '5': {'new_length': 217854252,
                   'old_length': 217853868,
                   'patches': [{'after': '16000000120000006f030000120000005f030000120000000f000000120000004c0300001200000053030000120000005d0300001200000076030000120000005203000012000000730300001200000055030000120000005603000012000000',
                                'before': '',
                                'offset': 54034780},
                               {'after': '4d0300001200000072030000120000005403000012000000140000001200000074030000120000007b0300001200000058030000120000005003000012000000780300001200000060030000120000005a030000120000005903000012000000',
                                'before': '',
                                'offset': 54629268},
                               {'after': '480300001200000013000000120000000c0000001200000010000000120000006e0300001200000011000000120000005e030000120000006c03000012000000150000001200000057030000120000006d030000120000000e00000012000000',
                                'before': '',
                                'offset': 154830272},
                               {'after': '59030000120000004d0300001200000072030000120000005403000012000000140000001200000074030000120000007b0300001200000058030000120000005003000012000000780300001200000060030000120000005a03000012000000',
                                'before': '',
                                'offset': 155424760}]},
             '6': {'new_length': 217854252,
                   'old_length': 217853868,
                   'patches': [{'after': '59030000590300005a0300005a03000060030000600300007803000078030000500300005003000058030000580300007b0300007b03000074030000740300001400000014000000540300005403000072030000720300004d0300004d030000',
                                'before': '',
                                'offset': 54034780},
                               {'after': '560300005603000055030000550300007303000073030000520300005203000076030000760300005d0300005d03000053030000530300004c0300004c0300000f0000000f0000005f0300005f0300006f0300006f0300001600000016000000',
                                'before': '',
                                'offset': 54629268},
                               {'after': '5a0300005a03000060030000600300007803000078030000500300005003000058030000580300007b0300007b03000074030000740300001400000014000000540300005403000072030000720300004d0300004d0300005903000059030000',
                                'before': '',
                                'offset': 154830272},
                               {'after': '0e0000000e0000006d0300006d030000570300005703000015000000150000006c0300006c0300005e0300005e03000011000000110000006e0300006e03000010000000100000000c0000000c00000013000000130000004803000048030000',
                                'before': '',
                                'offset': 155424760}]},
             '7': {'new_length': 1718572,
                   'old_length': 1718572,
                   'patches': [{'after': '16000000590300006f0300005a0300005f030000600300000f000000780300004c0300005003000053030000580300005d0300007b0300007603000074030000520300001400000073030000540300005503000072030000560300004d030000',
                                'before': '590300005a030000600300007803000050030000580300007b030000740300001400000054030000720300004d030000160000006f0300005f0300000f0000004c030000530300005d0300007603000052030000730300005503000056030000',
                                'offset': 115116},
                               {'after': '4d0300005603000072030000550300005403000073030000140000005203000074030000760300007b0300005d0300005803000053030000500300004c030000780300000f000000600300005f0300005a0300006f0300005903000016000000',
                                'before': '56030000550300007303000052030000760300005d030000530300004c0300000f0000005f0300006f030000160000004d030000720300005403000014000000740300007b030000580300005003000078030000600300005a03000059030000',
                                'offset': 115212},
                               {'after': '480300005a03000013000000600300000c0000007803000010000000500300006e03000058030000110000007b0300005e030000740300006c03000014000000150000005403000057030000720300006d0300004d0300000e00000059030000',
                                'before': '5a030000600300007803000050030000580300007b030000740300001400000054030000720300004d0300005903000048030000130000000c000000100000006e030000110000005e0300006c03000015000000570300006d0300000e000000',
                                'offset': 1481736},
                               {'after': '590300000e0000004d0300006d03000072030000570300005403000015000000140000006c030000740300005e0300007b03000011000000580300006e0300005003000010000000780300000c00000060030000130000005a03000048030000',
                                'before': '0e0000006d03000057030000150000006c0300005e030000110000006e030000100000000c0000001300000048030000590300004d030000720300005403000014000000740300007b030000580300005003000078030000600300005a030000',
                                'offset': 1481832}]},
             '8': {'new_length': 12203156,
                   'old_length': 12202964,
                   'patches': [{'after': '878b0100848b0100878b0100888b0100838b0100888b0100898b0100828b0100898b01008a8b0100818b01008a8b01008b8b0100808b01008b8b01008c8b01007f8b01008c8b01008d8b01007e8b01008d8b01008e8b01007d8b01008e8b01008f8b01007c8b01008f8b0100908b01007b8b0100908b0100918b01007a8b0100918b0100928b0100868b0100928b0100',
                                'before': '848b0100838b0100828b0100818b0100808b01007f8b01007e8b01007d8b01007c8b01007b8b01007a8b0100868b0100878b0100888b0100898b01008a8b01008b8b01008c8b01008d8b01008e8b01008f8b0100908b0100918b0100928b0100',
                                'offset': 9651688},
                               {'after': '868b0100928b0100868b01007a8b0100918b01007a8b01007b8b0100908b01007b8b01007c8b01008f8b01007c8b01007d8b01008e8b01007d8b01007e8b01008d8b01007e8b01007f8b01008c8b01007f8b0100808b01008b8b0100808b0100818b01008a8b0100818b0100828b0100898b0100828b0100838b0100888b0100838b0100848b0100878b0100848b0100',
                                'before': '928b0100918b0100908b01008f8b01008e8b01008d8b01008c8b01008b8b01008a8b0100898b0100888b0100878b0100868b01007a8b01007b8b01007c8b01007d8b01007e8b01007f8b0100808b0100818b0100828b0100838b0100848b0100',
                                'offset': 9655304},
                               {'after': '988b0100848b0100988b0100998b0100838b0100998b01009a8b0100828b01009a8b01009b8b0100818b01009b8b01009c8b0100808b01009c8b01009d8b01007f8b01009d8b0100978b01007e8b0100978b0100968b01007d8b0100968b0100958b01007c8b0100958b0100948b01007b8b0100948b0100938b0100928b0100938b0100868b01007a8b0100868b0100',
                                'before': '848b0100838b0100828b0100818b0100808b01007f8b01007e8b01007d8b01007c8b01007b8b0100928b01007a8b0100988b0100998b01009a8b01009b8b01009c8b01009d8b0100978b0100968b0100958b0100948b0100938b0100868b0100',
                                'offset': 11418420},
                               {'after': '7a8b0100868b01007a8b0100928b0100938b0100928b01007b8b0100948b01007b8b01007c8b0100958b01007c8b01007d8b0100968b01007d8b01007e8b0100978b01007e8b01007f8b01009d8b01007f8b0100808b01009c8b0100808b0100818b01009b8b0100818b0100828b01009a8b0100828b0100838b0100998b0100838b0100848b0100988b0100848b0100',
                                'before': '868b0100938b0100948b0100958b0100968b0100978b01009d8b01009c8b01009b8b01009a8b0100998b0100988b01007a8b0100928b01007b8b01007c8b01007d8b01007e8b01007f8b0100808b0100818b0100828b0100838b0100848b0100',
                                'offset': 11422036}]},
             '9': {'new_length': 15866708,
                   'old_length': 15866324,
                   'patches': [{'after': '878b0100848b0100888b0100838b0100898b0100828b01008a8b0100818b01008b8b0100808b01008c8b01007f8b01008d8b01007e8b01008e8b01007d8b01008f8b01007c8b0100908b01007b8b0100918b01007a8b0100928b0100868b0100',
                                'before': '',
                                'offset': 10453916},
                               {'after': '868b0100928b01007a8b0100918b01007b8b0100908b01007c8b01008f8b01007d8b01008e8b01007e8b01008d8b01007f8b01008c8b0100808b01008b8b0100818b01008a8b0100828b0100898b0100838b0100888b0100848b0100878b0100',
                                'before': '',
                                'offset': 10479900},
                               {'after': '988b0100848b0100998b0100838b01009a8b0100828b01009b8b0100818b01009c8b0100808b01009d8b01007f8b0100978b01007e8b0100968b01007d8b0100958b01007c8b0100948b01007b8b0100938b0100928b0100868b01007a8b0100',
                                'before': '',
                                'offset': 12472200},
                               {'after': '7a8b0100868b0100928b0100938b01007b8b0100948b01007c8b0100958b01007d8b0100968b01007e8b0100978b01007f8b01009d8b0100808b01009c8b0100818b01009b8b0100828b01009a8b0100838b0100998b0100848b0100988b0100',
                                'before': '',
                                'offset': 12498184}]}}}

FACTOR_SCHEMA = {'H1_I': {'descr': '<f4', 'fortran_order': False, 'shape': []},
 'H2_I': {'descr': '<f8', 'fortran_order': False, 'shape': [8, 44]},
 'Hsym_lower': {'descr': '<f4', 'fortran_order': False, 'shape': [2926]},
 'U': {'descr': '<f8', 'fortran_order': False, 'shape': [8, 143, 76]},
 'W': {'descr': '<f8', 'fortran_order': False, 'shape': [8, 143, 44]},
 'angle_words': {'descr': '<i8', 'fortran_order': False, 'shape': [1220, 75]},
 'tree_edges': {'descr': '<i8', 'fortran_order': False, 'shape': [75, 2]}}

"""Apply a finite, pinned packet patch to a public KMB circuit as data."""

import hashlib
import struct


def dqi_file_digest(path):
    with path.open("rb") as stream:
        return hashlib.file_digest(stream, "sha256").hexdigest()


def read_dqi_layout(stream):
    header = stream.read(104)
    assert header[:16].hex() == "d750c7d5c329d326e3cc9f6834f2b8bf"
    version, qubits, bits, registers = struct.unpack_from("<4I", header, 16)
    assert version == 2 and qubits == 892
    for _ in range(registers):
        tag, length = struct.unpack("<QQ", stream.read(16))
        assert tag == 0
        stream.seek(length, 1)
    register_end = stream.tell()
    packets = {}
    for expected_tag in range(1, 12):
        tag, length = struct.unpack("<QQ", stream.read(16))
        assert tag == expected_tag
        packets[tag] = (stream.tell(), length)
        stream.seek(length, 1)
    assert stream.tell() == stream.seek(0, 2)
    return header, register_end, packets


def copy_dqi_bytes(source, destination, count):
    assert count >= 0
    while count:
        block = source.read(min(count, 1 << 20))
        assert block
        destination.write(block)
        count -= len(block)


def apply_dqi_patch(baseline, destination, plan):
    assert dqi_file_digest(baseline) == plan["baseline_sha256"]
    with baseline.open("rb") as source, destination.open("xb") as result:
        header, register_end, packets = read_dqi_layout(source)
        assert header.hex() == plan["baseline_header"]
        result.write(bytes.fromhex(plan["candidate_header"]))
        source.seek(104)
        copy_dqi_bytes(source, result, register_end - 104)
        for tag, (start, length) in packets.items():
            entry = plan["packets"][str(tag)]
            assert length == entry["old_length"]
            result.write(struct.pack("<QQ", tag, entry["new_length"]))
            source.seek(start)
            cursor = 0
            output_bytes = 0
            for patch in entry["patches"]:
                offset = patch["offset"]
                copy_dqi_bytes(source, result, offset - cursor)
                before = bytes.fromhex(patch["before"])
                after = bytes.fromhex(patch["after"])
                assert source.read(len(before)) == before
                result.write(after)
                output_bytes += offset - cursor + len(after)
                cursor = offset + len(before)
            copy_dqi_bytes(source, result, length - cursor)
            assert output_bytes + length - cursor == entry["new_length"]
    assert dqi_file_digest(destination) == plan["candidate_sha256"]


"""Assemble the exact verified circuit and chemistry archive on Kaggle."""

import os
from pathlib import Path
import shutil
import tempfile
import zipfile

import numpy as np

INPUT = Path("/kaggle/input")
WORK = Path("/kaggle/working")


def member_hashes(archive):
    names = archive.namelist()
    assert len(names) == len(set(names)) == 18
    result = {}
    for name in names:
        with archive.open(name) as stream:
            result[name] = hashlib.file_digest(stream, "sha256").hexdigest()
    return result


def mounted_sources():
    paths = {}
    for path in INPUT.rglob("submission.zip"):
        checksum = dqi_file_digest(path)
        for name, source in SOURCES.items():
            if checksum == source["sha256"]:
                assert name not in paths
                paths[name] = path
    assert set(paths) == set(SOURCES)
    return paths


def check_original_source_hashes():
    roots = [p.parents[2] for p in INPUT.rglob("verify_submission_dqi.py")
             if all((p.parents[2] / n).is_file() for n in ORIGINAL_PINS)]
    assert len(roots) == 1
    assert {name: dqi_file_digest(roots[0] / name) for name in ORIGINAL_PINS} == ORIGINAL_PINS


def copy_member(archive, name, path):
    with archive.open(name) as source, path.open("xb") as dest:
        shutil.copyfileobj(source, dest)


def build_submission():
    check_original_source_hashes()
    mounted = mounted_sources()
    destination = WORK / "submission.zip"
    assert not destination.exists()
    with tempfile.TemporaryDirectory(dir=WORK) as directory:
        scratch = Path(directory)
        transport_dqi = scratch / "transport.kmb"
        dqi = scratch / "dqi.kmb"
        with zipfile.ZipFile(mounted["maoxin"]) as archive:
            assert member_hashes(archive) == SOURCES["maoxin"]["pins"]
            copy_member(archive, "dqi.kmb", transport_dqi)
            apply_dqi_patch(transport_dqi, dqi, TRANSPORT_PLAN)
            members = {"dqi.kmb": dqi}
            for name in sorted(PINS.keys() - members.keys()):
                path = scratch / name.replace("/", "__")
                copy_member(archive, name, path)
                assert dqi_file_digest(path) == PINS[name], name
                members[name] = path
        with np.load(members["dfthcsos/submission.npz"], allow_pickle=False) as factor:
            assert set(factor.files) == set(FACTOR_SCHEMA)
            for name, spec in FACTOR_SCHEMA.items():
                array = factor[name]
                assert array.dtype.str == spec["descr"] and list(array.shape) == spec["shape"], name
                if array.dtype.kind == "f":
                    assert np.isfinite(array).all(), name
        pending = scratch / "submission.zip"
        with zipfile.ZipFile(pending, "x") as archive:
            for name, path in sorted(members.items()):
                entry = zipfile.ZipInfo(name)
                entry.compress_type = zipfile.ZIP_DEFLATED
                with path.open("rb") as source, archive.open(entry, "w") as out:
                    shutil.copyfileobj(source, out)
        with zipfile.ZipFile(pending) as archive:
            assert member_hashes(archive) == PINS
        os.link(pending, destination)
    return destination
```
